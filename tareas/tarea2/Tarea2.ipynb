### Pregunta 1

```python
import numpy as np
```

```python
# Elegi 12 como semilla para que los resultados sean reproducibles

rng = np.random.default_rng(seed = 12)
```

```python
# Defino las luminosidades de las estrellas en unidades de luminosidad solar y las distancias en parsec

L = np.array([1.0, 2.5, 0.8, 5.0, 10.0])
d = np.array([10, 25, 15, 40, 60])
L, d
```

```python
# Constantes para convertir las unidades
L_solar = 3.828e26 # luminosidad solar a watts definido como L_solar
pc = 3.08e16 # Parsec definido como pc
def calcular_flujo(L, d):
    # Convierto la luminosidad solar a watts
    L = L * L_solar
    
    # Convierto la distancia de parsec a metros
    d = d * pc
    
    # Calculo el flujo observado
    F = L / (4 * np.pi * d**2)
    
    return F
F = calcular_flujo(L, d)

L, d, F
```

```python
# Indices de mayor o menor flujo
indice_mayor = np.argmax(F)

indice_menor = np.argmin(F)

indice_mayor, indice_menor
```

```python
# Promedio y mediana del flujo
flujo_promedio = np.mean(F)

flujo_mediana = np.median(F)

flujo_promedio, flujo_mediana
```

```python
# Defino el límite de detección
limite_deteccion = 1e-11

# Creo una máscara para las estrellas detectables
detectadas = F >= limite_deteccion

detectadas
```

```python
# Busco los índices de las estrellas detectadas
indices_detectadas = np.where(detectadas)[0]
indices_detectadas
```

```python
# Ordeno los índices según el flujo, de mayor a menor
flujos_ordenados = np.sort(F)[::-1]
flujos_ordenados
```

### Pregunta 2

```python
# Creo 500 tiempos equidistantes entre 0 y 10 años
t = np.linspace(0, 10, 500)

# Creo un flujo normalizado igual a 1
flujo = np.ones(500)

# Agrego ruido gaussiano
ruido = rng.normal(loc=0, scale=0.002, size=500)

# Agrego el ruido al flujo
flujo = flujo + ruido

# Identifico los puntos que están dentro del intervalo del tránsito
transito = (t >= 4.5) & (t <= 5.0)

# Disminuyo el flujo durante el tránsito
flujo[transito] = flujo[transito] - 0.02
```

```python
# Calculo el flujo promedio
flujo_promedio = np.mean(flujo)
flujo_promedio
```

```python
# Calculo la desviación estándar del flujo
desviacion = np.std(flujo)
desviacion
```

```python
# Busco el valor mínimo del flujo
flujo_minimo = np.min(flujo)

# Busco el índice donde ocurre el mínimo
indice_minimo = np.argmin(flujo)

flujo_minimo, indice_minimo
```

```python
# Busco el tiempo correspondiente al valor mínimo
tiempo_minimo = t[indice_minimo]

tiempo_minimo
```

```python
# Calculo el límite de 3 sigma
limite_3sigma = flujo_promedio - 3 * desviacion

# Selecciono los puntos que cumplen la condición
puntos_transito = flujo < limite_3sigma

puntos_transito
```

```python
# Cuento los puntos identificados como posibles puntos del tránsito
cantidad_puntos= np.sum(puntos_transito)

cantidad_puntos
```

```python
# Calculo el flujo promedio durante el tránsito
flujo_transito = np.mean(flujo[transito])

# Calculo la profundidad del tránsito
profundidad = 1 - flujo_transito

profundidad
```

```python
# Calculo la razón entre el radio del planeta y el radio de la estrella
razon_radios = np.sqrt(profundidad)

razon_radios
```

```python
# Obtengo los tiempos de los puntos identificados
tiempos_detectados = t[puntos_transito]

tiempos_detectados
```
